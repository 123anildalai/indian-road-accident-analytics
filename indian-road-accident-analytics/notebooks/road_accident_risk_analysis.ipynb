### Project Overview
This notebook provides an in-depth exploratory data analysis and risk modeling pipeline on 5,000+ Indian road accident incident records across states, highway categories, and temporal dimensions.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load preprocessed dataset
df = pd.read_csv("../data/cleaned_road_accidents.csv")
print(f"Dataset Shape: {df.shape}")
df.head()
```

## 1. Executive Statistical Overview
Evaluating total casualties, fatality ratios, and severity distributions.

```python
total_accidents = len(df)
total_fatalities = df["fatalities"].sum()
total_injuries = df["injuries"].sum()
fatality_rate = (total_fatalities / total_accidents) * 100

print(f"Total Crashes Recorded: {total_accidents:,}")
print(f"Total Lives Lost       : {total_fatalities:,}")
print(f"Total Persons Injured  : {total_injuries:,}")
print(f"National Fatality Rate : {fatality_rate:.2f}%")
```

## 2. Infrastructure Vulnerability: Highway vs Urban Roads

```python
road_summary = df.groupby("road_type").agg(
    incidents=("accident_id", "count"),
    fatalities=("fatalities", "sum"),
    overspeeding_share=("is_overspeeding", "mean")
).reset_index()
road_summary["fatality_rate_pct"] = (road_summary["fatalities"] / road_summary["incidents"] * 100).round(2)
road_summary["overspeeding_share"] = (road_summary["overspeeding_share"] * 100).round(1)
road_summary.sort_values(by="fatality_rate_pct", ascending=False)
```

## 3. High-Risk Temporal Spectrum (24-Hour Density)

```python
plt.figure(figsize=(10, 4.5))
hourly_data = df.groupby("hour")["accident_id"].count().reset_index()
sns.lineplot(data=hourly_data, x="hour", y="accident_id", marker="o", color="#d9534f", linewidth=2.5)
plt.fill_between(hourly_data["hour"], hourly_data["accident_id"], color="#d9534f", alpha=0.15)
plt.title("Accident Density Across 24 Hours", fontsize=13, fontweight="bold")
plt.xlabel("Hour of Day (00 to 23)")
plt.ylabel("Incident Count")
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```
